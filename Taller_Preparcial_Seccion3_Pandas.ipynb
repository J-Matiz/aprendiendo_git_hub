## Subsección 1: Limpieza de datos

**1.** La variable **Departamento** tiene valores faltantes (`NaN`) en las filas de **Juliana** y **Laura**. Escriba el código en pandas que impute estos valores utilizando la **moda** de la variable.

```python
for fila in df['Nombre']:
    print(f"Fila: {fila}")
    if fila == "Juliana" or fila == "Laura":
        print("Se encontró a Juliana o Laura, imputando")
        df.loc[df["Nombre"].isin(["Juliana", "Laura"]), "Departamento"] = (
            df["Departamento"].mode()[0]
        )
print(df[["Nombre", "Departamento"]])
```

**2.** La variable **AñosExperiencia** tiene un valor negativo en la fila de **Juliana** (`-3`), producto de un error de digitación. Escriba el código en pandas que corrija este valor, convirtiéndolo en positivo.

```python
print(df[df["Nombre"] == "Juliana"])
df["AñosExperiencia"] = df["AñosExperiencia"].apply(lambda x: abs(x))
print(df[df["Nombre"] == "Juliana"])
```

**3.** La variable **Genero** presenta inconsistencias: aparece registrada como `"Hombre"`, `"H"`, `"Mujer"` y `"M"`. Escriba el código en pandas que estandarice esta variable para que solo queden los valores `"Hombre"` y `"Mujer"`.

```python
df["Genero"] = df["Genero"].replace({"H": "Hombre", "M": "Mujer"})
df
```

**4.** El `DataFrame` contiene una fila **duplicada**: el empleado **Esteban** aparece registrado dos veces de forma idéntica. Escriba el código en pandas que detecte y elimine las filas duplicadas.

```python
df=df.drop_duplicates()
df
```

**5.** La variable **Salario** de **Catalina** tiene un valor atípico (`55.0`), muy alejado del resto de los datos. Corríjalo asumiendo que se trata de un error de digitación (un cero de más).

```python
df.loc[df["Nombre"] == "Catalina", "Salario"] = 5.5
df
```

**6.** El registro de **Manuela** está desactualizado: en realidad ella tiene **8 años** de experiencia. Escriba el código en pandas que corrija ese dato.

```python
df.loc[df["Nombre"] == "Manuela", "AñosExperiencia"] = 8
df
```

## Subsección 3: Agrupamiento (`groupby`)

Todas las preguntas de esta sección asumen que ya se creó la variable `Antiguedad` de la Subsección 2.

**1.** Calcule el **salario promedio** por `Departamento`, ordenado de mayor a menor.

```python
salario_promedio_departamento = (
    df.groupby("Departamento")["Salario"]
    .mean()
    .sort_values(ascending=False)
)
salario_promedio_departamento
```

**2.** Cuente cuántos empleados hay de cada `Genero`.

```python
empleados_por_genero = df["Genero"].value_counts()
empleados_por_genero
```

**3.** Encuentre el salario **máximo** y **mínimo** en cada `Ciudad`.

```python
salario_por_ciudad = df.groupby("Ciudad")["Salario"].agg(
    maximo="max",
    minimo="min"
)
salario_por_ciudad
```

**4.** Calcule, para cada categoría de `Antiguedad`, el promedio de **AñosExperiencia** y de **Salario** al mismo tiempo.

```python
resumen_antiguedad = df.groupby("Antiguedad", observed=True).agg(
    promedio_experiencia=("AñosExperiencia", "mean"),
    promedio_salario=("Salario", "mean")
)
resumen_antiguedad
```

**5.** Cuente cuántos empleados hay en cada combinación de `Departamento` y `Genero`.

```python
empleados_por_departamento_y_genero = (
    df.groupby(["Departamento", "Genero"]).size()
)
empleados_por_departamento_y_genero
```
